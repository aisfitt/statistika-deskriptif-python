```python
import pandas as pd
```

```python
# Tabel Data
path = 'aisyah.xlsx'
dataraw = pd.read_excel(path)
print(dataraw)
```

```python
# Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw['Grade'], columns='Frekuensi')
print(datafrq)
```

```python
import matplotlib.pyplot as plt

# Data dari tabel frekuensi
grades = datafrq.index.tolist()
freq = datafrq['Frekuensi'].tolist()
colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8172B2', '#937860']
```

```python
# Grafik Garis
plt.figure(figsize=(6,4))
plt.plot(grades, freq, marker='o', color='#4C72B0', linewidth=2)
plt.title('Distribusi Nilai (Grafik Garis)')
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.grid(alpha=0.3)
plt.show()
```

```python
# Grafik Batang
plt.figure(figsize=(6,4))
plt.bar(grades, freq, color=colors[:len(grades)])
plt.title('Distribusi Nilai (Grafik Batang)')
plt.xlabel('Grade')
plt.ylabel('Frekuensi')
plt.show()
```

```python
# Pie Chart
plt.figure(figsize=(6,6))
plt.pie(freq, labels=grades, autopct='%1.0f%%', colors=colors[:len(grades)], startangle=90)
plt.title('Distribusi Nilai (Pie Chart)')
plt.show()
```

```python
# Statistika Deskriptif
dataraw["NILAI"] = pd.to_numeric(dataraw["NILAI"], errors='coerce')
dt = dataraw["NILAI"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print(stats)
```
